## Primeiro contato, pelas três portas de entrada


```python
from pyspark.sql import SparkSession

spark = (SparkSession.builder
         .appName("aula02")
         .master("local[*]")
         .getOrCreate())

print(spark.version)
print(spark.sparkContext.master)
```

```python
spark.range(5).selectExpr("id", "id * 10 AS dez").show()
```
